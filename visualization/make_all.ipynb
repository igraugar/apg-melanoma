## Figures

Each PDF is shown at screen resolution. The manuscript uses the PDF files themselves. Figure 1 is a schematic whose rankings, arrows and support values illustrate the method and are not read from the results.

```python
import pymupdf
from IPython.display import Image, Markdown, display

FIGURES = [
    ("Figure 1", "fig0_pipeline"),
    ("Figure 2", "fig1_reliance"),
    ("Figure 3", "fig2_ambiguity"),
    ("Figure 4", "fig3_pairs"),
    ("Figure 5", "fig4_exchanges_tiers"),
    ("Figure 6", "fig5_seeds"),
    ("Figure 7", "fig6_global_local"),
]

for label, name in FIGURES:
    page = pymupdf.open(f"figures/{name}.pdf")[0]
    display(Markdown(f"**{label}** (`{name}.pdf`)"))
    display(Image(data=page.get_pixmap(dpi=60).tobytes("png")))
```
